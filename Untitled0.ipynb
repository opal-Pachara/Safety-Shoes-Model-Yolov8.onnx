```python
import wandb
import torch

from ultralytics import YOLO, settings

print("Ultralytics loaded")
print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print(
        "VRAM:",
        round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2),
        "GB"
    )
```

```python
settings.update({
    "wandb": True
})

print("Ultralytics W&B:", settings["wandb"])
```

```python
wandb.init(
    project="yolo12-Safety_Shoes",
    name="Yolo12n-safety_shoes",
    config={
        "model": "YOLO12-N",
        "imgsz": 640,
        "batch": 4,
        "workers": 2,
        "optimizer": "AdamW",
        "lr0": 0.001,
        "lrf": 0.01,
        "weight_decay": 0.0005,
        "epochs": 100,
        "patience": 50,
    }
)

print("W&B Run:", wandb.run.name)
print("W&B URL:", wandb.run.url)
```

```python
model = YOLO(
    "yolo12n.pt"
)

model.info()
```

```python
def log_wandb(trainer):

    if wandb.run is None:
        return

    epoch = trainer.epoch + 1
    metrics = trainer.metrics

    log_data = {
        "epoch": epoch,

        "train/box_loss": float(trainer.tloss[0]),
        "train/cls_loss": float(trainer.tloss[1]),
        "train/dfl_loss": float(trainer.tloss[2]),

        "val/mAP50": metrics.get("metrics/mAP50(B)", 0),
        "val/mAP50-95": metrics.get("metrics/mAP50-95(B)", 0),
        "val/precision": metrics.get("metrics/precision(B)", 0),
        "val/recall": metrics.get("metrics/recall(B)", 0),
    }

    wandb.log(log_data)
```

```python
model.add_callback(
    "on_fit_epoch_end",
    log_wandb
)

print("W&B callback registered")
```
